```python
import pandas as pd

# Cargar el archivo original
df = pd.read_csv("BDisney.csv")

# ---------------------------
# ARCHIVO DE PELÍCULAS
# ---------------------------
movies_cols = [
    "title", "plot", "rated", "year", "released_at", "runtime",
    "genre", "director", "writer", "actors", "language", "country",
    "awards", "metascore", "imdb_rating", "imdb_votes"
]

df_movies = df[df["type"] == "movie"][movies_cols]
df_movies.to_csv("movies.csv", index=False)

# ---------------------------
# ARCHIVO DE SERIES
# ---------------------------
series_cols = [
    "title", "plot", "rated", "year_start", "year_end", "released_at", "runtime",
    "genre", "director", "writer", "actors", "language", "country",
    "awards", "metascore", "imdb_rating", "imdb_votes"
]

df_series = df[df["type"] == "series"][series_cols]
df_series.to_csv("series.csv", index=False)

print("Archivos generados correctamente:")
print("- movies.csv")
print("- series.csv")
```

```python
# Para evitar problemas por nombres repetidos separados por comas:
# Convertimos cada fila en una lista de directores
df["director_list"] = df["director"].str.split(",")

# Expandimos para tener una fila por director individual
df_exploded = df.explode("director_list")
df_exploded["director_list"] = df_exploded["director_list"].str.strip()

# Función para obtener lista de títulos filtrados
def titles_by_type(group, tipo):
    return group[group["type"] == tipo]["title"].dropna().unique().tolist()

# Agrupamos por director
director_df = df_exploded.groupby("director_list").apply(
    lambda g: pd.Series({
        "name": g.name,
        "films": titles_by_type(g, "movie"),
        "tvshows": titles_by_type(g, "series")
    })
).reset_index(drop=True)

# Exportar a CSV
director_df.to_csv("directors.csv", index=False)

print("Archivo generado: directors.csv")
```

```python
# Para evitar problemas por nombres repetidos separados por comas:
# Convertimos cada fila en una lista de directores
df["writer_list"] = df["writer"].str.split(",")

# Expandimos para tener una fila por director individual
df_exploded = df.explode("writer_list")
df_exploded["writer_list"] = df_exploded["writer_list"].str.strip()

# Función para obtener lista de títulos filtrados
def titles_by_type(group, tipo):
    return group[group["type"] == tipo]["title"].dropna().unique().tolist()

# Agrupamos por director
director_df = df_exploded.groupby("writer_list").apply(
    lambda g: pd.Series({
        "name": g.name,
        "films": titles_by_type(g, "movie"),
        "tvshows": titles_by_type(g, "series")
    })
).reset_index(drop=True)

# Exportar a CSV
director_df.to_csv("writers.csv", index=False)

print("Archivo generado: writers.csv")
```

```python
import pandas as pd

# Cargar tu archivo
df = pd.read_csv("disney.csv")

# Columnas que deben convertirse de string → array
cols = ["films", "shortFilms", "tvShows"]

def convert_to_array(x):
    if pd.isna(x) or x.strip() == "":
        return []       # manejar vacíos
    return [i.strip() for i in x.split(",")]  # lista limpia

# Convertir cada una
for c in cols:
    df[c] = df[c].apply(convert_to_array)

# Guardar como CSV nuevamente
df.to_csv("disney_arrays.csv", index=False)

print("Archivo disney_arrays.csv generado con arrays listos para MongoDB")
```
